```python
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer

# 1. Clean the text and convert comma-separated strings into clean Python lists
def clean_addon_string(text):
    # Handle NaN or empty string inputs gracefully
    if pd.isna(text) or text == '':
        return []
    # Split by comma and strip out accidental whitespace around names
    return [addon.strip() for addon in text.split(',') if addon.strip()]

# Apply the text cleaning function to your column
df_car['addons_list_clean'] = df_car['addons_list'].apply(clean_addon_string)

# 2. Initialize the MultiLabelBinarizer with Sparse Output enabled
# This completely prevents memory crashes by ignoring zero values
mlb = MultiLabelBinarizer(sparse_output=True)

# 3. Fit and transform the clean lists into a compressed matrix
X_addons_sparse = mlb.fit_transform(df_car['addons_list_clean'])

# 4. Verify your results
print(f"Compressed Sparse Matrix Shape: {X_addons_sparse.shape}")
print(f"Detected Unique Addon Columns: {mlb.classes_}")
```

```python
sample_df = pd.DataFrame(
    X_addons_sparse[:5].toarray(), 
    columns=mlb.classes_
)
sample_df
```

```python
# 1. Convert the sparse matrix into a standard DataFrame with proper column names
addons_df = pd.DataFrame.sparse.from_spmatrix(
    X_addons_sparse, 
    columns=[f"addon_{name.lower().replace(' ', '_')}" for name in mlb.classes_]
)

# 2. Reset indices to ensure perfect horizontal alignment
df_car = df_car.reset_index(drop=True)
addons_df = addons_df.reset_index(drop=True)

# 3. Glue them together side-by-side
df_car = pd.concat([df_car, addons_df], axis=1)
```

```python
from sklearn.feature_selection import mutual_info_regression

# 1. Temporarily fill missing text values for calculation
X_cat_temp = df_car[category].fillna('Unknown')

# 2. Convert text words into numerical codes so the MI algorithm can read them
for col in category:
    X_cat_temp[col] = X_cat_temp[col].astype('category').cat.codes

# 3. Calculate Mutual Information scores
mi_scores = mutual_info_regression(X_cat_temp, df_car['annual_premium'], random_state=42)

# 4. Sort and display the rankings
cat_ranking = pd.DataFrame({
    'Categorical Feature': category,
    'Mutual Info Score': mi_scores
}).sort_values(by='Mutual Info Score', ascending=False)

print("\n🏆 Categorical Columns Ranked by Relationship to Premium:")
print(cat_ranking.to_string(index=False))
```

```python
num = [
    'customer_age', 'city_tier', 'city_risk_score', 'manufacturing_year', 
    'vehicle_age_years', 'engine_cc', 'idv', 'ncb_percent', 
    'claim_history_count', 'num_addons',
    'addon_consumables_cover', 'addon_engine_protection', 'addon_key_replacement', 
    'addon_no_claim_bonus_protection', 'addon_passenger_cover', 'addon_personal_accident_cover',
    'addon_return_to_invoice', 'addon_roadside_assistance', 'addon_tyre_protection', 
    'addon_zero_depreciation'
]

# 2. CORRECT CATEGORICAL LIST
cat = [
    'vehicle_make', 'vehicle_model', 'variant', 'segment', 
    'fuel_type','policy_type',
    'city', 'state'
]
```

```python
X = df_car[num + cat]
Y = df_car['annual_premium']
```

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, TargetEncoder, StandardScaler,LabelEncoder,OrdinalEncoder
from xgboost import XGBRegressor
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
cat_low = ['segment', 'fuel_type', 'policy_type']
cat_high = ['city', 'state', 'vehicle_make', 'vehicle_model', 'variant']
base_numeric = ['customer_age', 'city_tier', 'city_risk_score', 'vehicle_age_years', 
                'engine_cc', 'idv', 'ncb_percent', 'claim_history_count', 'num_addons','manufacturing_year']
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        # Encode ALL text categories safely with zero memory overhead
        ('all_categorical_encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), 
         ['segment', 'fuel_type', 'policy_type', 'city', 'state', 'vehicle_make', 'vehicle_model', 'variant']),
        
        # Scale continuous numbers
        ('numerical_scaler', StandardScaler(), 
         ['customer_age', 'city_tier', 'city_risk_score', 'vehicle_age_years', 
          'engine_cc', 'idv', 'ncb_percent', 'claim_history_count', 'num_addons','manufacturing_year']),
    ],
    remainder='passthrough',
    sparse_threshold=0.0
)
```

```python
X_train_final = preprocessor.fit_transform(X_train, y_train)
X_test_final = preprocessor.transform(X_test)
```

```python
model = XGBRegressor(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    tree_method='hist', # Super-fast optimization for heavy datasets
    random_state=42,
    n_jobs=-1
)

model.fit(X_train_final, y_train)
print("🏆 Model training complete!")
```

```python
print(f"Final Model Train R² Score: {model.score(X_train_final, y_train):.4f}")
print(f"Final Model Test R² Score: {model.score(X_test_final, y_test):.4f}")
```
